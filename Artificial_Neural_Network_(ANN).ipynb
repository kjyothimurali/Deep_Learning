#XOR Gate Problem using ANN forward propagation only
Architecture:
2 Inputs -> 2 Hidden Neurons -> 1 Output

```python
import numpy as np
X=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
    ])
y=np.array([0,1,1,0])
X,y
```

#Define Step Activation Function

```python
#activation function
def activation_function(z):
    return 1 if z>=0 else 0
```

#Initialize weights and biases (Manually Choosen)


```python
#Hidden layer weights and biases
w1,w2,b1=1,1,-0.5
w3,w4,b2=1,1,-1.5
#output layer weights and bias
w5,w6,b3=1,-2,-0.5
w=[w1,w2,w3,w4,w5,w6]
b=[b1,b2,b3]
print("Initial weight:",w)
print("Initial bias:",b )
```

#Define forward propagation function

```python
#Froward Propagation
def forward_propagation(x1,x2):
  h1=w1*x1+w2*x2+b1
  a_h1=activation_function(h1)
  h2=w3*x1+w4*x2+b2
  a_h2=activation_function(h2)

  y=w5*a_h1+w6*a_h2+b3
  output=activation_function(y)
  return output
```

#Test Forward Propagation for all XOR Inputs


```python
print("XOR Results using ANN Forward Propagation")
for x in X:
  prediction=forward_propagation(x[0],x[1])
  print(f'Input: {x} Output: {prediction}')
```

#Plotting XOR Decision Boundary

```python
import numpy as np
import matplotlib.pyplot as plt

# XOR Dataset
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0,1,1,0])

# ---- Activation functions ----
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# ---- Manually chosen weights that solve XOR ----
W1 = np.array([[20,20],
               [20,20]])
b1 = np.array([-10, -30])

W2 = np.array([[20],
               [-20]])
b2 = np.array([-10])

# ---- Forward Propagation ----
def forward(X):
    Z1 = np.dot(X, W1) + b1
    A1 = sigmoid(Z1)

    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)

    return A2

# ---- Create meshgrid for decision boundary ----
xx, yy = np.meshgrid(np.linspace(-0.5,1.5,200),
                     np.linspace(-0.5,1.5,200))

grid = np.c_[xx.ravel(), yy.ravel()]
pred = forward(grid)
pred = pred.reshape(xx.shape)

# ---- Plot decision boundary ----
plt.figure(figsize=(5,4))

plt.contourf(xx, yy, pred > 0.5, alpha=0.3)

# Plot XOR points
plt.scatter(X[y==0][:,0], X[y==0][:,1],
            color='blue', s=150, edgecolors='k', label='Class 0')

plt.scatter(X[y==1][:,0], X[y==1][:,1],
            color='red', s=150, edgecolors='k', label='Class 1')

plt.title("XOR Decision Boundary (2 Hidden Neurons)")
plt.xlabel("Input Feature 1")
plt.ylabel("Input Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```
