#Perceptron (Binary Classification)
### example: study hours->result

```python
import numpy as np
```

Prepare Training Data

```python
X=np.array([1,2,3,4])
y=np.array([0,0,1,1])
X, y
```

Initialize Parameters

```python
w=0.0
b=0.0
learning_rate=1
epochs=5
print("Initial weight:",w)
print("Initial bias:",b)
```

Define Activation Function (Step Function)

```python
def step_function(z):
  return 1 if z>=0 else 0
```

Training Loop

```python
for epoch in range(epochs):
  print(f"\nEpoch: {epoch+1}")
  for i in range(len(X)):
    #1. Compute weighted sum
    z=w*X[i]+b
    #2. Apply Activation function
    y_pred=step_function(z)

    #3. Compare the error
    error=y[i]-y_pred

    #4. Update Weight
    w=w+learning_rate*error*X[i]

    #5. Update bias
    b=b+learning_rate*error

    print(f"Input:{X[i]}, Predicted: {y_pred}, Actual: {y[i]}, Weight: {w}, Bias: {b}")
```

Final Parameters

```python
print("\n final Weight:",w)
print("\n final bias:",b)
```

Testing Model

```python
test_val=3.1
z=w*test_val+b
prediction=step_function(z)
print(f"For study hours {test_val}, Prediction is {prediction}")
```

```python
import matplotlib.pyplot as plt
#plotting with decision boundry
plt.scatter(X,y,color='red')
plt.plot(X,w*X+b  ,color='blue')
plt.xlabel('Study Hours')
plt.ylabel('Result')
plt.title('Perceptron Learning')
plt.grid(True)
plt.show()
```

Perceptron AND Gate

```python
X=np.array([
    [0,0],[0,1],[1,0],[1,1]
])
y=np.array([0,0,0,1]) #AND Output
X, y
```

```python
w=np.zeros(2)
b=0
learning_rate=0.1
epochs=10
print("Initial weight:",w)
print("Initial bias:",b )
```

```python
def step_function(z):
  return 1 if z>=0 else 0
```

```python
for epoch in range(epochs):
  print(f"\nEpoch: {epoch+1}")
  for i in range(len(X)):
    #1. Compute weighted sum
    z=np.dot(w,X[i])+b
    #2. Apply Activation function
    y_pred=step_function(z)

    #3. Compare the error
    error=y[i]-y_pred

    #4. Update Weight
    w=w+learning_rate*error*X[i]

    #5. Update bias
    b=b+learning_rate*error

    print(f"Input:{X[i]}, Predicted: {y_pred}, Actual: {y[i]}, Weight: {w}, Bias: {b}")
```

```python
print("\n final Weight:",w)
print("\n final bias:",b)
```

```python
print("\n Testing AND Gate")
for i in range(len(X)):
  z=np.dot(w,X[i])+b
  prediction=step_function(z)
  print(f"Input:{X[i]}, Predicted: {prediction}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Get the min/max values for features to set plot limits
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5

# Create a range of x values for plotting the decision boundary
line_x = np.linspace(x_min, x_max, 100)

# Calculate the corresponding y values for the decision boundary
# Equation of the decision boundary: w[0]*x1 + w[1]*x2 + b = 0
# Solve for x2: x2 = (-w[0]*x1 - b) / w[1]
if w[1] != 0:
    line_y = (-w[0] * line_x - b) / w[1]
else:
    # If w[1] is 0, the boundary is a vertical line: w[0]*x1 + b = 0 => x1 = -b/w[0]
    line_y = np.linspace(y_min, y_max, 100)
    line_x = np.full_like(line_y, (-b / w[0]) if w[0] != 0 else 0)

# Plot the input data points, colored by their class (y)
plt.scatter(X[:,0], X[:,1], c=y, cmap='viridis', marker='o', s=100, edgecolors='k', label='Data Points')

# Plot the decision boundary
plt.plot(line_x, line_y, color='red', linestyle='-', label='Decision Boundary')

plt.xlabel('Input Feature 1')
plt.ylabel('Input Feature 2')
plt.title('Perceptron for AND Gate with Decision Boundary')
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.grid(True)
plt.legend()
plt.show()
```
